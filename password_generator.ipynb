```python
# Password Generator 
# Password length should be between 8 to 64
# Password must contain one upper_case, one lower_case , one digit and one special character

import secrets, string
length = int (input ('Please enter the length of password : '))
password = ''
if 8 <= length <= 64:
  char = string.ascii_letters + string.digits + string.punctuation
  while True:

    for i in range (length):
      password+=secrets.choice(char)
    if any (i.isupper() for i in password) and any (i.islower() for i in password) and any (i.isdigit() for i in password) and any (i in string.punctuation for i in password):
      break
    else:
      password = ''
else:
  print ('Invalid Length')
print (password)
```
